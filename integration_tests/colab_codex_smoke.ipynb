```python
from statistics import mean
from collections import Counter

# Datos estrictamente sinteticos; no se leen archivos privados.
datos = [
    {'id': 1, 'valor': 100},
    {'id': 2, 'valor': 120},
    {'id': 3, 'valor': 110},
    {'id': 4, 'valor': 130},
    {'id': 2, 'valor': None},
    {'id': 5, 'valor': 120},
]
print('Dataset sintetico cargado:', len(datos), 'filas')
```

```python
ids = [x['id'] for x in datos]
duplicados = sorted(k for k, n in Counter(ids).items() if n > 1)
valores = [x['valor'] for x in datos if x['valor'] is not None]
reporte = {
    'filas': len(datos),
    'ids_unicos': len(set(ids)),
    'ids_duplicados': duplicados,
    'valores_faltantes': sum(x['valor'] is None for x in datos),
    'promedio_no_nulos': mean(valores),
}
print(reporte)
```

```python
assert reporte['filas'] == 6
assert reporte['ids_unicos'] == 5
assert reporte['ids_duplicados'] == [2]
assert reporte['valores_faltantes'] == 1
assert reporte['promedio_no_nulos'] == 116
print('SMOKE TEST PASS: ejecucion de notebook, conteo, duplicados, nulos y promedio')
```
